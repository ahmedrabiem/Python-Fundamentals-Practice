```python
print(datetime.datetime(1982, 10, 25))
print(datetime.datetime(1982, 10, 25, 10, 45, 55, 150364))
```

```python
birthday=datetime.datetime(2004,6,14,12,23,13,12321)
date=datetime.datetime.now()

print(f"My Birthday is {birthday} And ", end="")
print(f"Date Now Is {date}")
```

```python
print(f"I lived for {birthday - date}")
print(f"I lived for {(birthday - date).days} day")
```


##### -- Date and Time => Format Date --

###### https://strftime.org/


```python

import datetime

myBirthday = datetime.datetime(2004, 6, 14)

print(myBirthday)
print(myBirthday.strftime("%a"))
print(myBirthday.strftime("%A"))
print(myBirthday.strftime("%b"))
print(myBirthday.strftime("%B"))

print(myBirthday.strftime("%d %B %Y"))
print(myBirthday.strftime("%d, %B, %Y"))
print(myBirthday.strftime("%d/%B/%Y"))
print(myBirthday.strftime("%d - %B - %Y"))
print(myBirthday.strftime("%B - %Y"))
```
