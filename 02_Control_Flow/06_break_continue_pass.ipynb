### Break, Continue, Pass

## continue 
##### بلغي الخطوه اللي حددتها وبكمل بعدها

```python
myrange = range(1,11)
for num in myrange:
    if num == 4:
        continue
    print(num)
```

## Break
##### بوقف عند الخطوه اللي عايزها

```python
myrange = range(1,11)
for num in myrange:
    if num == 4:
        break
    print(num)
```

```python
myrange = range(1,11)
for num in myrange:
    print(num)
    if num == 4:
        break
    
```

## Pass
##### لو في ايرور بعديه

```python
myrange = range(1,11)
for num in myrange:
    if num == 4:
        pass #لو لغيت الكومنت مش هيكلع ايرور
    print(num)
```
