```python
myfweb=[]
max = 5
print("you can add 5 web: ")

while max >0 :
    x = input("do you want to continue?yes or no")
    if x == "yes":
        web=input("enter web without https://")
        myfweb.append(f"https:// {web.strip().lower()}")
        max -=1
        print(f"website added,{max} places left")
        print (f"your F Web :{myfweb}")
    if x == "no":
        print("end")
        break
else:
    print("you can't add more")
```

```python
mylist = []
a = 6

while a > 0:
    admin = input("Enter admin name: ").strip().title()  # لتحسين تنسيق الأسماء
    mylist.append(admin)  # إضافة الاسم إلى القائمة
    print(f"You have {a - 1} places left")  # التصحيح هنا ليظهر العدد المتبقي بعد إدخال الاسم
    print(f"Admin name added: {admin}")
    a -= 1
else:
    print("Your list is full.")

# ترتيب القائمة أبجدياً
mylist.sort()

# عرض قائمة الإداريين
index = 0
if len(mylist) > 0:
    print("\nList of admins:")
    while index < len(mylist):
        print(f"#{str(index + 1).zfill(2)} {mylist[index]}")  # الترقيم المنسق
        index += 1
```

```python
tries = 5
truepas="ahmed&"
inputpas = input("enter passwored : ")

while inputpas != truepas :
    tries-=1
    print(f"wrong pas ,{"last" if tries==0 else tries} chance")
    inputpas = input("enter passwored : ")

    if tries == 0:
        print("All tries are finished.")
        break
else:
    print("correct pas")
```

```python
tries = 3  # عدد المحاولات المتاحة

while tries >= 0:
    password = input("Enter your password: ")
    if password == "ahmed&":
        print("Access granted!")
        break
    else:
        tries -= 1
        print(f"Incorrect password. {"last"if tries==0 else tries} tries left.")
        
    if tries == 0:
        print("All tries are finished.")
        break
```
