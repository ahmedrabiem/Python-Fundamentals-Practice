### Nested If

```python
name = input("enter your Fullname: ").title().strip()
country = input("enter your country: ")
student = input("Are you a student?")
price = 100
cname = "montage"

if country=="Egypt" or country== "Kuwait":
    print(f"Hello {name} because you are from {country}")
    if student=="yes":
        print(f"the course {cname} price is : ${price- 90}")
    else:
        print(f"the course {cname} price is : ${price-80}")
elif country == "KSA":
    print(f"Hello {name} because you are from {country}")
    print(f"the course {cname} price is : ${price- 60}")
else :
    print(f"Hello {name} because you are from {country}")
    print(f"the course {cname} price is : ${price}")
```
