```python

from pathlib import Path
import re
import shutil

# Change this path to your actual basics folder.
BASE_DIR = Path(r"D:\almgal\python pr\python")

# Keep True to preview changes without modifying files.
DRY_RUN = True

# Set True to move unrecognized files into an "Unsorted" folder.
MOVE_UNSORTED = False

# Each folder has a list of filename keywords.
# More specific rules should appear before broader rules.
CATEGORIES = {
    "01_Basics_and_Data_Types": [
        "str", "string", "tuple", "bool", "set", "dictionary",
        "comparison of data types", "assignment operator",
        "comparison operator", "type conversion", "user input",
        "slice email", "membership", "zip", "pillow"
    ],
    "02_Conditions": [
        "if elif else", "nested if", "ternary if"
    ],
    "03_Loops_and_Iteration": [
        "while loop", "break,continue,pass", "for loop",
        "iterator vs iterable"
    ],
    "04_Functions": [
        "func type", "function", "recursion", "lambda",
        "built in func", "decorator", "doc string"
    ],
    "05_Modules_and_DateTime": [
        "modules", "module.py", "date time"
    ],
    "06_Errors_and_Debugging": [
        "errors and exceptions", "try, except, else, finally",
        "debugging"
    ],
    "07_File_Handling": [
        "file handling"
    ],
    "08_Practice_and_Projects": [
        "calculator", "chatbot", "decode the message", "rev"
    ],
}

# Files are sorted by their existing numeric prefix first.
def natural_sort_key(path):
    name = path.stem.lower()
    match = re.match(r"^\s*(\d+(?:\.\d+)*)\s*", name)

    if match:
        numbers = tuple(
            int(part) for part in match.group(1).split(".")
        )
        return (0, numbers, name)

    return (1, (), name)


def normalize(text):
    return re.sub(r"[\s_.-]+", " ", text.lower()).strip()


def find_category(filename):
    name = normalize(Path(filename).stem)

    # Remove the old numeric prefix for matching.
    name = re.sub(r"^\d+(?:\.\d+)*\s*", "", name).strip()

    for folder, keywords in CATEGORIES.items():
        for keyword in keywords:
            if normalize(keyword) in name:
                return folder

    return "Unsorted"


def main():
    if not BASE_DIR.exists():
        print(f"Folder not found: {BASE_DIR}")
        print("Update BASE_DIR to your actual folder path.")
        return

    files = [
        path for path in BASE_DIR.iterdir()
        if path.is_file()
        and path.name != Path(__file__).name
    ]

    groups = {}

    for path in files:
        category = find_category(path.name)

        if category == "Unsorted" and not MOVE_UNSORTED:
            print(f"[SKIP] {path.name} - no category matched")
            continue

        groups.setdefault(category, []).append(path)

    operations = []

    for category, paths in groups.items():
        paths.sort(key=natural_sort_key)

        for number, source in enumerate(paths, start=1):
            target_dir = BASE_DIR / category
            target = target_dir / (
                f"{number:02d}_{source.name}"
            )

            # Remove old numbering from the filename.
            clean_name = re.sub(
                r"^\s*\d+(?:\.\d+)*\s*", "", source.name
            ).strip()

            target = target_dir / f"{number:02d}_{clean_name}"

            if target.exists() and target.resolve() != source.resolve():
                print(f"[CONFLICT] {target}")
                print("Skipped to avoid overwriting a file.")
                continue

            operations.append((source, target))

    print("\nPlanned changes:")
    for source, target in operations:
        print(f"{source.name} -> {target.parent.name}/{target.name}")

    if DRY_RUN:
        print("\nDRY RUN: No files were changed.")
        print("Review the plan, then set DRY_RUN = False.")
        return

    # Phase 1: Move files to temporary names to avoid collisions.
    staged = []

    for index, (source, target) in enumerate(operations):
        target.parent.mkdir(parents=True, exist_ok=True)
        temporary = source.with_name(
            f"__organizing_temp_{index}__{source.suffix}"
        )

        if temporary.exists():
            raise FileExistsError(temporary)

        shutil.move(str(source), str(temporary))
        staged.append((temporary, target))

    # Phase 2: Move staged files to their final destinations.
    for temporary, target in staged:
        shutil.move(str(temporary), str(target))

    print(f"\nDone! Organized {len(staged)} files.")


if __name__ == "__main__":
    main()
```

```python

from pathlib import Path
import re

# ==================================================
# 1. SETTINGS
# ==================================================

BASE_DIR = Path(r"D:\almgal\python pr\python\basics")

# True = preview only
# False = actually move and rename files
DRY_RUN = True

# ==================================================
# 2. HELPER FUNCTIONS
# ==================================================

def clean_filename(filename):
    """Remove an existing numeric prefix from a filename."""
    return re.sub(
        r"^\s*\d+(?:\.\d+)*\s*",
        "",
        filename
    ).strip()


def normalize_filename(filename):
    """Normalize a filename for category matching."""
    name = Path(filename).stem.lower()
    name = clean_filename(name)

    name = re.sub(r"[_-]+", " ", name)
    name = re.sub(r"\s+", " ", name)

    return name.strip()


def sort_key(path):
    """Sort numbered lessons before unnumbered lessons."""
    match = re.match(
        r"^\s*(\d+(?:\.\d+)*)",
        path.stem
    )

    if match:
        numbers = tuple(
            int(part)
            for part in match.group(1).split(".")
        )
        return (0, numbers, path.name.lower())

    return (1, (), path.name.lower())


# ==================================================
# 3. FILE CLASSIFICATION
# ==================================================

def find_category(path):
    """Return the category folder for a file."""
    name = normalize_filename(path.name)

    # File handling
    if "file handling" in name:
        return "08_File_Handling"

    # Conditions
    if any(term in name for term in [
        "if,elif,else",
        "nested if",
        "ternary if"
    ]):
        return "02_Conditions"

    # Loops
    if any(term in name for term in [
        "while loop",
        "for loop",
        "break,continue,pass",
        "break, continue, pass"
    ]):
        return "03_Loops"

    # Functions
    if any(term in name for term in [
        "func type",
        "function",
        "recursion",
        "lambda",
        "bult in fonc",
        "built in func",
        "built in fonc"
    ]):
        return "04_Functions"

    # Errors and debugging
    if any(term in name for term in [
        "errors and exceptions",
        "try, except",
        "try,except",
        "debugging"
    ]):
        return "07_Errors_and_Debugging"

    # Modules and dates
    if any(term in name for term in [
        "modules",
        "module",
        "date time"
    ]):
        return "05_Modules_and_DateTime"

    # Advanced Python
    if any(term in name for term in [
        "iterator",
        "iterable",
        "decorator",
        "zip",
        "pillow",
        "doc string",
        "commenting",
        "documenting"
    ]):
        return "06_Advanced_Python"

    # Basic data types and operators
    if any(term in name for term in [
        "string",
        "str",
        "tuple",
        "bool",
        "set",
        "dictionary",
        "comparison of data types",
        "assignment operator",
        "comparison operator",
        "type conversion",
        "user input",
        "slice email",
        "membership"
    ]):
        return "01_Basics_and_Data_Types"

    # Practice and projects
    if any(term in name for term in [
        "calculator",
        "chatbot",
        "decode the message",
        "rev",
        "webscrabing",
        "amt"
    ]):
        return "09_Practice_and_Projects"

    # Files that need manual review
    return "Unsorted"


# ==================================================
# 4. BUILD THE ORGANIZATION PLAN
# ==================================================

def build_plan():
    """Create a plan without moving any files."""

    if not BASE_DIR.is_dir():
        raise FileNotFoundError(
            f"Folder does not exist: {BASE_DIR}"
        )

    # Only process files directly inside BASE_DIR.
    files = [
        path
        for path in BASE_DIR.iterdir()
        if path.is_file()
        and not path.name.startswith("__temp_organize_")
    ]

    groups = {}

    for path in files:
        category = find_category(path)
        groups.setdefault(category, []).append(path)

    operations = []

    for category, paths in groups.items():
        paths.sort(key=sort_key)

        for number, source in enumerate(paths, start=1):
            new_name = clean_filename(source.name)

            target_dir = BASE_DIR / category
            target = target_dir / f"{number:02d}_{new_name}"

            operations.append((source, target, category))

    return files, operations


# ==================================================
# 5. DISPLAY THE PLAN
# ==================================================

def show_plan(files, operations):
    """Display planned moves and possible conflicts."""

    print("=" * 65)
    print("PYTHON FILE ORGANIZER")
    print("=" * 65)

    conflicts = []

    for source, target, category in operations:
        print(f"\nFile: {source.name}")
        print(f"Category: {category}")
        print(f"Destination: {target.parent.name}\\{target.name}")

        # Ignore an existing target if it is the same file.
        if target.exists() and target.resolve() != source.resolve():
            conflicts.append(target)
            print("WARNING: Destination already exists!")

    unsorted_count = sum(
        1 for _, _, category in operations
        if category == "Unsorted"
    )

    print("\n" + "=" * 65)
    print(f"Files discovered: {len(files)}")
    print(f"Files in plan: {len(operations)}")
    print(f"Unsorted files: {unsorted_count}")
    print(f"Conflicts: {len(conflicts)}")
    print("=" * 65)

    return conflicts, unsorted_count


# ==================================================
# 6. EXECUTE THE PLAN SAFELY
# ==================================================

def execute_plan(operations):
    """Move and rename files after checking destinations."""

    # Do not proceed if any destination is occupied.
    for source, target, category in operations:
        if target.exists() and target.resolve() != source.resolve():
            raise FileExistsError(
                f"Destination already exists: {target}"
            )

    # Stage files in temporary paths to avoid rename collisions.
    staged = []

    try:
        for index, (source, target, category) in enumerate(operations):
            temporary = BASE_DIR / f"__temp_organize_{index}__"

            if temporary.exists():
                raise FileExistsError(
                    f"Temporary path already exists: {temporary}"
                )

            source.rename(temporary)
            staged.append((temporary, target))

        # Move each staged file to its final destination.
        for temporary, target in staged:
            target.parent.mkdir(parents=True, exist_ok=True)
            temporary.rename(target)

        print(f"\nSuccess! Organized {len(staged)} files.")

    except Exception:
        print("\nAn error occurred during file organization.")
        print("Checking whether staged files can be restored...")

        # Restore files that have not reached their final destination.
        for temporary, target in reversed(staged):
            if temporary.exists():
                original = next(
                    (
                        source
                        for source, planned_target, _
                        in operations
                        if planned_target == target
                    ),
                    None
                )

                if original is not None and not original.exists():
                    original.parent.mkdir(
                        parents=True, exist_ok=True
                    )
                    temporary.rename(original)

        raise


# ==================================================
# 7. MAIN
# ==================================================

def main():
    files, operations = build_plan()

    conflicts, unsorted_count = show_plan(files, operations)

    if DRY_RUN:
        print("\nDRY RUN: No files have been moved.")
        print("Review the plan and your backup first.")
        print("If correct, set DRY_RUN = False.")
        return

    if conflicts:
        print("\nSTOP: Resolve destination conflicts first.")
        return

    if unsorted_count:
        print("\nSTOP: Some files are not classified.")
        print("Add category rules before moving any files.")
        return

    confirmation = input(
        "\nType YES to move and rename the files: "
    ).strip()

    if confirmation != "YES":
        print("Cancelled. No files were moved.")
        return

    execute_plan(operations)


main()
```

##### Comparison of Data Types and Their Methods in Python

#### Integer Numbers (int)
###### Description: Integer numbers (e.g., 1, 2, 100).
###### Mutable: No (Immutable).
###### Methods: No specific methods, but supports arithmetic operations 
###### such as addition, subtraction, multiplication, division, and so on.

#### Floating-Point Numbers (float)
###### Description: Decimal numbers (e.g., 3.14, 2.71).
###### Mutable: No (Immutable).
###### Methods: No specific methods, but supports arithmetic operations 
###### and mathematical functions (from the math library).


##### *****************************************

#### Strings (str)
###### Description: Text values (e.g., "hello").
###### Mutable: No (Immutable).
###### Methods:
###### Examples include:
###### - capitalize(): Capitalizes the first letter.
###### - lower(): Converts all characters to lowercase.
###### - upper(): Converts all characters to uppercase.
###### - replace(old, new): Replaces occurrences of a substring.
###### - split(delimiter): Splits the string into a list.
###### - strip(): Removes leading and trailing whitespace.
###### Many more methods exist for advanced string manipulation.


##### *****************************************

#### Lists (list)
###### Description: A collection of elements (e.g., [1, 2, 3]).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - append(element): Adds an element to the end of the list.
###### - clear(): Removes all elements from the list.
###### - pop(index): Removes and returns the element at the specified index.
###### - sort(): Sorts the list in place.
###### - reverse(): Reverses the order of elements in the list.
###### - remove(element): Removes the first occurrence of the element.


##### *****************************************

#### Tuples (tuple)
###### Description: Immutable collection of elements (e.g., (1, 2, 3)).
###### Mutable: No (Immutable).
###### Methods:
###### - count(element): Counts the occurrences of an element.
###### - index(element): Returns the index of the first occurrence of an element.


##### *********************************************
#### Dictionaries (dict)
###### Description: A collection of key-value pairs (e.g., {"name": "Ahmed", "age": 20}).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - keys(): Returns a view of all keys in the dictionary.
###### - values(): Returns a view of all values in the dictionary.
###### - items(): Returns a view of all key-value pairs.
###### - update(new_dict): Updates the dictionary with another dictionary.
###### - pop(key): Removes and returns the value for the given key.
###### - clear(): Removes all key-value pairs from the dictionary.


#### Dictionaries (dict)
###### Description: A collection of key-value pairs (e.g., {"name": "Ahmed", "age": 20}).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - keys(): Returns a view of all keys in the dictionary.
###### - values(): Returns a view of all values in the dictionary.
###### - items(): Returns a view of all key-value pairs.
###### - update(new_dict): Updates the dictionary with another dictionary.
###### - pop(key): Removes and returns the value for the given key.
###### - clear(): Removes all key-value pairs from the dictionary.


##### ********************************************

#### Sets (set)
###### Description: Unordered collection of unique elements (e.g., {1, 2, 3}).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - add(element): Adds an element to the set.
###### - remove(element): Removes an element from the set (raises an error if the element is not found).
###### - discard(element): Removes an element from the set (does not raise an error if the element is not found).
###### - union(other_set): Returns a set containing all unique elements from both sets.
###### - intersection(other_set): Returns a set of common elements between two sets.
###### - difference(other_set): Returns a set of elements in the first set but not in the second.
###### - isdisjoint(other_set): Checks if two sets have no elements in common.
