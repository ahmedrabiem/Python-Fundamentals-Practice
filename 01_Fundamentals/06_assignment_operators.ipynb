### ------------------------
### -Assignment Operators-
### ------------------------

##### (=) / (+=) / (-=) / (*=)
##### (/=) / (**=) / (%=) / (//=)

```python
x = 10 # var1
y = 20 # var2
# var1 = var1 [operator +] var2
x = x +y

# var1 [operator +]= var2
z = 10 # var1
v = 20 # var2
z+=v
print(x)
print(z)
```

```python
x = 10 # var1
y = 20 # var2
# var1 = var1 [operator **] var2
x = x ** y

z = 10 # var1
v = 20 # var2
# var1 [operator **]= var2
z**=v
print(x)
print(z)
```
