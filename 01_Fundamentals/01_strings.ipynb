#### 1. تعديل السلسلة النصية
##### .lower()
###### Converts the string to lowercase.
###### Example: "Hello".lower() → "hello"

##### .upper()
###### Converts the string to uppercase.
###### Example: "Hello".upper() → "HELLO"

##### .capitalize()
###### Capitalizes only the first letter and makes the rest lowercase.
###### Example: "hello".capitalize() → "Hello"

##### .title()
###### Capitalizes the first letter of each word.
###### Example: "hello world".title() → "Hello World"

##### .swapcase()
###### Swaps the case of all characters (uppercase to lowercase and vice versa).
###### Example: "Hello".swapcase() → "hELLO"

#### 2. البحث في السلسلة النصية
##### .find(substring)
###### Finds the first occurrence of a substring and returns its index, or -1 if not found.
###### Example: "hello".find("e") → 1

##### .index(substring)
###### Similar to .find() but raises an error if the substring is not found.
###### Example: "hello".index("e") → 1

##### .rfind(substring)
###### Finds the last occurrence of a substring.
###### Example: "hello".rfind("l") → 3

##### .startswith(substring)
###### Checks if the string starts with the specified substring.
###### Example: "hello".startswith("he") → True

##### .endswith(substring)
###### Checks if the string ends with the specified substring.
###### Example: "hello".endswith("lo") → True

#### 3. تعديل محتوى السلسلة النصية
##### .strip()
###### Removes leading and trailing spaces or specified characters.
###### Example: "  hello  ".strip() → "hello"

##### .lstrip()
###### Removes leading spaces or specified characters.
###### Example: "  hello".lstrip() → "hello"

##### .rstrip()
###### Removes trailing spaces or specified characters.
###### Example: "hello  ".rstrip() → "hello"

##### .replace(old, new)
###### Replaces occurrences of a substring with another substring.
###### Example: "hello".replace("l", "x") → "hexxo"

#### 4. تقسيم السلسلة النصية
##### .split(delimiter)
###### Splits the string into a list using a specified delimiter.
###### Example: "a,b,c".split(",") → ["a", "b", "c"]

##### .rsplit(delimiter)
###### Splits the string into a list starting from the right.
###### Example: "a,b,c".rsplit(",", 1) → ["a,b", "c"]

##### .splitlines()
###### Splits the string into a list by line breaks.
###### Example: "hello\nworld".splitlines() → ["hello", "world"]

##### .partition(substring)
###### Splits the string into 3 parts based on the substring.
###### Example: "hello world".partition(" ") → ("hello", " ", "world")

##### .rpartition(substring)
###### Splits the string starting from the last occurrence of the substring.
###### Example: "hello world".rpartition(" ") → ("hello", " ", "world")

#### 5. التحقق من السلسلة النصية
##### .isalnum()
###### Checks if the string contains only letters and numbers.
###### Example: "hello123".isalnum() → True

##### .isalpha()
###### Checks if the string contains only letters.
###### Example: "hello".isalpha() → True

##### .isdigit()
###### Checks if the string contains only digits.
###### Example: "123".isdigit() → True

##### .islower()
###### Checks if all characters in the string are lowercase.
###### Example: "hello".islower() → True

##### .isupper()
###### Checks if all characters in the string are uppercase.
###### Example: "HELLO".isupper() → True

##### .isspace()
###### Checks if the string contains only spaces.
###### Example: "   ".isspace() → True

##### .istitle()
###### Checks if the string is in title case (first letter of each word is capitalized).
###### Example: "Hello World".istitle() → True

#### 6. تنسيق السلسلة النصية
##### .center(width)
###### Centers the string within a specified width by padding.
###### Example: "hello".center(10) → "  hello   "

##### .ljust(width)
###### Left-aligns the string within a specified width by padding.
###### Example: "hello".ljust(10) → "hello     "

##### .rjust(width)
###### Right-aligns the string within a specified width by padding.
###### Example: "hello".rjust(10) → "     hello"

##### .zfill(width)
###### Pads the string with zeros from the left to reach a specified width.
###### Example: "42".zfill(5) → "00042"

##### .format()
###### Inserts values into the string using curly braces {}.
###### Example: "Hello {}".format("Ahmed") → "Hello Ahmed"

#### 7. الطباعة والتكرار
##### 'separator'.join(iterable)
###### Joins a list of strings using a specified separator.
###### Example: ",".join(["a", "b", "c"]) → "a,b,c"

##### .count(substring)
###### Counts the number of occurrences of a substring in the string.
###### Example: "hello".count("l") → 2
