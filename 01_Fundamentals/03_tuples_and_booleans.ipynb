### Tuple

#### Tuple شبه الليست في حجات كتير ولكن
##### 1- وجود الاقواس مش مهم فيها وبستخدم () Parentheses
##### 2- عناصر مرتبه اقدر استخدم index علشان اوصل لعنصر محدد
##### 3- Immutable => يعني متقدرش تضيف او تمسح عناصر
##### 4- تقدر تحط اي نوع بيانات انت عايزه نفس الليست
##### 5- بتستخدم العمليات عليها عادي


# ####Tuple Method####

```python
mytuple="ahmed","rabie","sayed"
my=("ahmed","rabie","sayed")
print(type(mytuple))
print(type(my))
print(my)
```

```python
a = (1,2,3,4,"sayed","Ahmed")
print(a[-1])
print(a[3])
```

```python
#مش بتقبل اي تعديل
a = (1,2,3,4,"sayed","Ahmed")
a[3]="Ahmed"
print(a)
```

#### tuple لو عايز اعرف عنصر واحد ك

```python
#بحط (,) بعد العنصر
a = "Ahmed",
x = ("Ahmed")
print(type(a))
print(type(x))
```

### بتستخدم العمليات عليها عادي نفس الليست و الاسترينج

```python
x=(1,2,3,4,"s")
z = 3,4,7
print(x+z)
c= x+ ("ahmed","rabie") +z
print(c)

s=[1,2,3,4,"s"]
b = [3,4,7]
print(s+b)
```

### Repeat بتمشي علي string ,tuple,list (*)

```python
mystring="ahmed"
mytuple= ("mohamed","salem")
mylist = [1,2,3]
print(mystring*3)
print(mytuple*3)
print(mylist*3)
```

### count

```python
a= (2,3,4,1,2,3,4,2,2)

print(a.count(2))
```

### index

```python

x = (1,2,3,4,1,2,3)

# print("the position of the index is: " + x.index(3)) #Error علشان مينفعش اضيف سترينج علي tuple

print("the position of the index is: {:d}".format(x.index(3)) )

print(f"the position of the index is: {x.index(3)}")
```

### Tuple Destruct علشان اخزن العناصر بتعتي في variable

```python
a = ("A","B","C")
x,y,z = a
print(x)
print(y)
print(z)
```

```python
#لو عايز اتجاهل عنصر معين
a = ("A","B","C","D")
x,y,_,z=a
print(x)
print(y)
print(z)
```

# True values

```python
print(bool("as"))
print(bool(100))
print(bool(True))
print(bool(10.2))
print(bool([1,3,4,4]))
```

# False values

```python

print(bool(""))
print(bool())
print(bool(0))
print(bool(False))
print(bool([]))
print(bool(None))
```

# and

```python
age=20
country = "egypt"
name = "Ahmed"

print (age > 19 and name == "Ahmed" and country =="egypt")

print("*"*10)

age=20
country = "egypt"
name = "Ahmed"

print (age > 30 and name == "Ahmed" and country =="egypt")
```

# or

```python

age=20
country = "egypt"
name = "Ahmed"

print (age > 30 or name == "Ahmed" or country =="KSA")
print("*"*10)

age=20
country = "egypt"
name = "Ahmed"

print (age > 30 or name == "Sayed" or country =="KSA")
```
