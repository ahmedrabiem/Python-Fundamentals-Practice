### List method

#### ###append###هو هنا ضاف الليست كامله ك عنصر

```python

mylist=["ahmed","Rabie","Mohamed"]
add=["safa","Abdalla"]
mylist.append(add)
print(mylist)
###لو عايز عنصر معين من الليست التنايه
print(mylist[3][1])
```

#### ###extend### هي هي نفس ال append بس دي بتاخد كل عنصر من الليس المضافه لوحده

```python

mylist=["ahmed","Rabie","Mohamed"]
add=["safa","Abdalla"]
mylist.extend(add)
print(mylist[3])
print(mylist)
```

#### ## Sort ## هنل برتب الليست بتعتي ولازم يكونو من نفس النوع يعني بيرتب سترينج عادي

```python

s=['S','X','A','c','B']
s.sort()
print(s)

a=[1,3,5,7,2,4,6,8,9,0]
a.sort()
print(a)
#لو عايز اعمل ترتيب معكوس
a.sort(reverse = True)
print(a)
##############################
# a.sort(reverse=False)
a.sort(reverse=False)
print(a)
```

#### ###reverse ### دي بتعكس الاسترينج

```python

mylist=["ahmed",2,4,"Rabie",7,"Mohamed"]
mylist.reverse()
print(mylist)
```

##### Remove هنا بحذف عنصر من الليست ولو متكرر بحذف اول واحد

```python
##### Remove هنا بحذف عنصر من الليست ولو متكرر بحذف اول واحد
mylist=["Ahmed","Rabie","Mohamed","Ahmed"]
mylist.remove("Ahmed")
print(mylist)
```

#### ### POP بطلع عنصر معين من الليست ب الاندكس بتاعه

```python
a=[1,2,3,4,2,3,4]
print(a.pop(2))
print(a.pop(2))
print(a)
```

#### clear بحذف كل عناصر الليست بتعتي

#### insert() بضيف عنصر في مكان محدد قبل المكان اللي حددته ال append بضيف ف الاخر بس

#### copy باخد نسخه من الليست ومش بتتأثر بالعمليات اللي علي الاصليه
###### shallow copy

```python
# ###copy باخد نسخه من الليست ومش بتتأثر بالعمليات اللي علي الاصليه
#shallow copy
mylist=["ahmed",2,4,"Rabie",7,"Mohamed"]
c=mylist.copy()
print(c)
mylist.append(5)
print(mylist)
print(c)
```
