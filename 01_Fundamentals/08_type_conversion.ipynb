#### tuple()

```python
string1 = "Ahmed" 
list1 = [1,2,3,4]
set1 = {"A","B","C"}
Dict1 = {"A":1 , "B":2}

print (tuple(string1))
print(tuple(list1))
print(tuple(set1))
print(tuple(Dict1)) # خد ال key
```

#### List()

```python
string1 = "Ahmed" 
tuple1 = (1,2,3,4)
set1 = {"A","B","C"} 
Dict1 = {"A":1 , "B":2}

print (list(string1))
print (list(tuple1))
print (list(set1))
print (list(Dict1))
```

### set()

```python
string1 = "Ahmed" 
tuple1 = (1,2,3,4)
list1 = [1,2,3,4]
Dict1 = {"A":1 , "B":2}

print (set(string1))
print (set(tuple1))
print (set(list1))
print (set(Dict1))
set(list1).pop()
print (set(list1).pop())
print (set(list1))
# لما حولت الليست الي ست كل مره كان بعمل ست جديده علشان كدا لما طبعت في الاخر طبع بدون حذف عنصر
# لو عايز احذف لازم احفظ الست ودا بيمشي في كل الداتا تيب
my_set = set(list1)
my_set.pop()
print(my_set)
```
