# Union 

```python

set1 = {3,'ahmed'}
set2 ={110,2,"asd",4,5}
set3 = {"ah", "df"}

print(set1 | set2 | set3)

print(set1.union (set2,set3))
```

#### symmetric_diffrance() المجموعه فرق تقاطعها مع المجموعتين التانيه


```python
a={1,2,3,5,24,4}
b={1,2,9,3}
print(a)
print(a.symmetric_difference(b)) #a^b
print(a^b)
print(a)
```

### superset() جزئيه وغير جزئيه


```python
a={1,2,3,5,24,4}
b={1,2,3}
print(a)
print(a.issuperset(b)) #a>=b
print(a>=b)
print(a)
```

# issubset


```python
a={1,2,3,5,24,4}
b={1,2,3,34}
print(a)
print(a.issubset(b)) #a<=b
print(a<=b)
print(a)
```

#### isdisjoin بشوف في تقاطع ولا لا


```python
a={1,2,3,5,24,4}
b={1,2,3}
c={12,10,11}
print(a)
print(a.isdisjoint(b))
print(a.isdisjoint(c)) 

print(a)
```

#### [2] Dict key has only immutable data type (number,string,tuple) list and dict are not
#### immutable => يمكن التعديل عليها


```python
user ={
    "name" : "ahmed",
    "age" : 20,
    "lan" : "python",
    (1,2,3) : "test",
    1 : "ads"
}
print(user)
```

### [3] Dict value can have any data type


```python

user ={
    "name" : "ahmed",
    "age" : 20,
    "lan" : "python",
    (1,2,3) : "test",
    1 : ["ahmed", "Rabie" ]
}
print(user)
```

#### [4] Dict key need to be unique


```python

user ={
    "name" : "ahmed",
    "name" : "sayed",
    "age" : 20,
    "lan" : "python",
    (1,2,3) : "test",
    1 : "ads"
}
print(user)
```

### [5] Dict is not orderd you access its element with key


```python
user ={
    "name" : "sayed",
    "age" : 20,
    "lang" : "python",
    (1,2,3) : "test",
    1 : "ads"
}
print(user)

print(user["name"])
print(user.get("name"))

print(user.keys())
print(user.values())
print(user.items())
```

#### TOW-DIMENSIONAL DICT


```python
user ={
    "name" : "sayed",
    "age" : 20,
    "info": {
        "fcolor":"black",
        "fcar" : "BMW"
    },
    "sayed" : {
        "weight" : 70,
        "length" : 180,
    }
}
print(user)

print(user["info"])
print(user["sayed"])

print(user["info"]["fcolor"])
# Dict len
print(len(user))
print(len(user["info"]))
```

```python
info= {
    "fcolor":"black",
    "fcar" : "BMW"
}

sayed = {
    "name" : "sayed",
    "weight" : 70,
    "length" : 180,
    }
learn ={
    "colleg" : "computer scince"
}

person={
    "one" : info,
    "two": sayed,
    'three': learn
}
print(person)
print(person["one"]["fcar"])
```
