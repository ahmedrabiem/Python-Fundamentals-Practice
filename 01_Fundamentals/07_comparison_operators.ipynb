#### Comparison Operator

#### [==] Equal
#### [!=] Not Equal

```python
print (100 == 100)
print (100 == 123)
print("*" * 10)
print (100 != 100)
print (100 != 123)
```

##### [>] Greater Than
##### [<] Less Than

```python
print (100 > 100)
print (100 > 123)
print("*" * 10)
print (100 < 100)
print (100 < 123)
```

##### [>=] Greater Than Or Equal
##### [<=] Less Than Or Equal

```python
print (100 >= 100)
print (100 >= 123)
print("*" * 10)
print (100 <= 100)
print (100 <= 123)
```
