#### الفرق بين استخدام `return` وعدم استخدامه في الدوال:

---

### بدون استخدام `return`:
- **الإرجاع**:  
  لا يتم إرجاع أي قيمة (القيمة الافتراضية هي `None`).

- **الاستخدام لاحقًا**:  
  لا يمكن استخدام الناتج أو تخزينه في متغير، لأنه غير موجود.

- **السلوك المتوقع**:  
  تُستخدم الدالة لتنفيذ مهام معينة فقط (مثل الطباعة أو تنفيذ إجراءات) دون الحاجة إلى إرجاع نتائج.

- **متى تُستخدم؟**  
  عندما تريد أن تنفذ الدالة إجراءً دون الحاجة إلى حفظ أو استخدام أي ناتج لاحقًا.

---

### مع استخدام `return`:
- **الإرجاع**:  
  يتم إرجاع قيمة محددة إلى المكان الذي استُدعيت منه الدالة.

- **الاستخدام لاحقًا**:  
  يمكن استخدام الناتج الذي تم إرجاعه أو تخزينه في متغير لاستخدامه لاحقًا.

- **السلوك المتوقع**:  
  تُستخدم للحصول على نتائج من العمليات أو الحسابات التي تنفذها الدالة.

- **متى تُستخدم؟**  
  عندما تحتاج إلى استخدام الناتج من الدالة في مكان آخر أو تخزينه.

# الفرق بين Function و Procedure و Parameter:

---

### <small>Function:</small>
- **التعريف**:  
  <sub>دالة (Function) تُستخدم لتنفيذ عملية أو حساب معين، وتُرجع دائمًا قيمة أو نتيجة عند استدعائها.</sub>

- **الإرجاع**:  
  <sub>تُرجع قيمة باستخدام `return`.</sub>

- **الاستخدام**:  
  <sub>تُستخدم غالبًا لإجراء حسابات أو عمليات تنتج عنها قيمة تُستخدم لاحقًا.</sub>

- **متى تُستخدم؟**  
  <sub>عندما تحتاج إلى معالجة بيانات أو القيام بعملية حسابية، مع أهمية استخدام الناتج في مكان آخر.</sub>

---

### <small>Procedure:</small>
- **التعريف**:  
  <sub>إجراء (Procedure) يُستخدم لتنفيذ مجموعة من التعليمات، لكنه لا يُرجع قيمة عند استدعائه (إلا إذا كان ذلك غير مباشر).</sub>

- **الإرجاع**:  
  <sub>لا يتم إرجاع أي قيمة (القيمة الافتراضية هي `None`).</sub>

- **الاستخدام**:  
  <sub>تُستخدم لتنفيذ مهام معينة، مثل الطباعة أو تحديث قاعدة بيانات أو إجراء تغييرات.</sub>

- **متى تُستخدم؟**  
  <sub>عندما تحتاج إلى تنفيذ مجموعة من العمليات دون الحاجة إلى نتيجة مباشرة.</sub>

---

### <small>Parameter:</small>
- **التعريف**:  
  <sub>معطى (Parameter) هو قيمة أو متغير يتم تمريره إلى الدالة أو الإجراء لتوفير مدخلات أثناء التنفيذ.</sub>

- **أنواع البراميتر**:  
  <sub>1. **Parameters**: يتم تعريفها في تعريف الدالة أو الإجراء.  
  2. **Arguments**: هي القيم الفعلية التي يتم تمريرها عند استدعاء الدالة.</sub>

- **الاستخدام**:  
  <sub>تُستخدم لتخصيص العمليات داخل الدالة أو الإجراء بناءً على المدخلات.</sub>

- **أمثلة**:  
  <sub>تمرير عددين لدالة جمع لإجراء عملية حسابية بناءً على هذه القيم.</sub>

#### Packing and Unpacking argument

```python
def hello(*peoples): # type => tuple
    for n in peoples:
        print(f"hello {n}")
        
hello("as","ahmed","sayed")
```

```python
def skills(name , *skill): # type => tuple
    print(f"hello {name}")
    for k in skill:
        print(f"your skills is => {k}")
name=input("PLS ENTER YOUR NAME :")
skills(name,"css","js")
```

#### Packing and Unpacking KeyWord argument 

```python
user ={
    "fname" : "ahmed",
    "lname" : "sayed",
    "age" : 20,
    "lan": ("python","css"),
    # (1,2,3) : "test", => لو لغيت الكمنت هيطلع ايرور دي بتشتغل علي سترينج بس
    # 1 : "ads"
}
def showskills(**skills): # type => dictionary
    for k,v in skills.items():
        print(f"{k} => {v}")

showskills(html="50%",css="70%")
showskills(**user)
```
