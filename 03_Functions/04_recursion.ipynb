### Function Recursion

```python
def sum_numbers(n):
    # الحالة الأساسية
    if n == 0:
        return 0
    # الاستدعاء الذاتي
    return n + sum_numbers(n - 1)

# تجربة الدالة
print(sum_numbers(5))  # الناتج: 15
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
p = int(input("pls enter num: "))
print(factorial(p))
```

```python
def clean(word):
    if len(word)==1:
        return word
    if word[0]==word[1]:
        return clean(word[1:])
    return word[0] + clean(word[1:])

print(clean("wwwooorrlldd"))
```

```python
def exponentiation(base,exponent):
    if exponent == 0:
        return 1
    
    return base*exponentiation(base,exponent-1)
    

print(exponentiation(2,3))
    
```

```python
def reverse(word):
    if len(word)==1:
        return word
    return word[-1] + reverse(word[:-1])

print(reverse("hello"))
```

```python
def lis(*l):
    if len(l)==0:
        return 0
    return len(l)
print(lis(3,4,5,5,3,2))
```

```python
def count_elements(lst):
    # الحالة الأساسية: إذا كانت القائمة فارغة
    if len(lst) == 0:
        return 0
    # التكرار الذاتي: إزالة العنصر الأول واستدعاء الدالة على باقي القائمة
    return 1 + count_elements(lst[1:])

# تجربة الدالة
print(count_elements([3, 4, 5, 5, 3, 2]))  # الناتج: 6
print(count_elements([]))                 # الناتج: 0
```

```python
def sumoflist(n):
    if len(n) ==0:
        return 0
    return n[0] + sumoflist (n[1:])

print(sumoflist([1,3,4,5]))
print(sumoflist([]))
```

```python
def count_even_numbers(l):
    if len(l)==0:
        return 0
    if l[0]%2==0:
        return 1+ count_even_numbers(l[1:])
    else:
        return count_even_numbers(l[1:])
    
print(count_even_numbers([1,2,3,4,5,6]))
```
