#### FUNC Scope

```python
x=1
def one():
    global x  #######
    x=4
    print(f"var x = {x}")
    
def two():
    x=10
    print(f"var x = {x}")
    
def three():
    print(f"var x = {x}")
    
print(f"var x ={x}")
one()
print(f"var x ={x}")
two()
three()
```

```python
n = int(input("Pls enter num: "))
factorial = 1  # البداية بقيمة 1 لأن الفاكتوريال عملية ضرب

if n == 0 or n == 1:  # الحالة الخاصة للـ 0 و 1
    print(f"The factorial of {n} is 1")
else:
    while n > 1:  # استمر بالضرب حتى تصل إلى 1
        factorial *= n  # اضرب العدد الحالي في الفاكتوريال
        n -= 1  # قلل العدد بمقدار 1
    print(f"The factorial is: {factorial}")
```
