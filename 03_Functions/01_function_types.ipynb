```python
# 📌 أنواع الـ Functions في بايثون

# 🔹 1. Built-in Functions (الدوال المدمجة)
# دوال جاهزة في بايثون مثل print(), len(), sum()
print(len([1, 2, 3]))  # مثال على دالة مدمجة

# 🔹 2. Predefined Functions (الدوال المعرفة مسبقًا)
# دوال تأتي مع مكتبات مثل math.sqrt(), random.randint()
import math
print(math.sqrt(16))  # مثال على دالة من مكتبة math

# 🔹 3. User-Defined Functions (الدوال المعرفة بواسطة المستخدم)
# دوال يكتبها المستخدم باستخدام def
def greet(name):
    return f"Hello, {name}!"

print(greet("Ahmed"))  # مثال على دالة معرفة بواسطة المستخدم

# 🔹 4. Lambda Functions (الدوال المجهولة)
# دوال قصيرة تستخدم lambda
square = lambda x: x ** 2
print(square(4))  # مثال على دالة Lambda

# 🔹 5. Recursive Functions (الدوال العودية)
# دوال تستدعي نفسها لحل المشاكل المتكررة مثل حساب الفاكتوريل
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))  # مثال على دالة عودية

# 🔹 6. Functions with Default Parameters (دوال تحتوي على بارامترات افتراضية)
# دوال تحتوي على بارامترات بقيم افتراضية
def greet(name="Guest"):
    return f"Hello, {name}!"

print(greet())       # مثال عند عدم تمرير بارامتر
print(greet("Ali"))  # مثال عند تمرير بارامتر
```
