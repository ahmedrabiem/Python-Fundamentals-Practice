
##### -- Built In Functions --

##### [1] all => بشوف لو كل العناصر true

```python
a =[1,2,3,4,5]
if all(a):
    print("all element is true")
else:
    print("at least one element is false")

x =[0,2,3,4,[]]
if all(x):
    print("all element is true")
else:
    print("at least one element is false")
```

##### [2] any => بشوف لو في عنصر واحد true

```python
a =[1,2,3,4,5]
if any(a):
    print("at least one element is true")
else:
    print("all element is false")

a =[0,0,[]]
if any(a):
    print("at least one element is true")
else:
    print("all element is false")
```

##### [3]bin=> بجيب البيناري بتاع الرقم

```python
print(bin(100))
```

##### [4] id => بجيب مكان العنصر في الميموري

```python
a = 1
b = 2

print(id(a))
print(id(b))
```

##### [5] sum(iterable,start)

```python
a=[1,3,4,5,6]
print(sum(a))
print(sum(a,40))
```

##### [6] round(number, num of digits) => بقرب الرقم

```python
print(round(150.55))
print("*"*5)
print(round(150.45))
print("*"*5)
print(round(150.455,2))
print("*"*5)
print(round(150.45,2))
```

##### [7] range(start مش اجباري,end اجباري, step مش اجباري)

```python
l=range(10)
print(*l)
```

```python
print(list(range(0)))
print(list(range(10)))
print(list(range(0,21,2)))
```

##### [8] print

```python
print("hallo ahmed how are you")
print("hallo" ,"ahmed",'how','are','you')
```

```python
print("hallo ahmed how are you")
print("hallo" ,"ahmed",'how','are','you',sep="@")
print("hallo" ,"ahmed",'how','are','you',sep="|")
```

```python
print("f line")
print("s line")
print("f line" ,end="|") # end => \n اساسي
print("s line")
```

##### [9] abs() => القيمه المطلقه

```python
print(abs(100))
print(abs(-100))
print(abs(-100.2))
```

##### [10] pow(base, exp, mod مش اجباري) => اس

```python
print(pow(2,5)) #2*2*2*2*2
print(pow(2,5,3)) #2*2*2*2*2 %3
```

##### [11] min(item,item,item,item, or iterator => حاجه يتعلم عليها لوب)

```python
print(min(15,12,3,43,2,-23))
print(min("a","x","b","sayed"))
print(min("z","x","y","sayed"))
```

```python
mytuple=(1,2,3,4,5,-1)
print(min(mytuple))
```

##### [12] max(item,item,item,item, or iterator => حاجه يتعلم عليها لوب)

```python
print(max(15,12,3,43,2,-23))
print(max("a","x","b","sayed"))
print(max("z","x","y","sayed"))
```

```python
mytuple=(1,2,3,4,5,-1)
print(max(mytuple))
```

##### [13] slice()

```python
a=["a","x","b","d","s"]
print("1",a[:])
print("2",a[:3])
print("3",a[1:3])
print("4",a[slice(4)])
print("5",a[slice(2,4)])
print("6",a[slice(0,5,2)])
```

#### enumerate(iterable, start=0)

```python
mySkills = ["Html", "Css", "Js", "PHP"]
mySkills_co = enumerate(mySkills,3)
print(mySkills_co)
for n, i in mySkills_co:
    print(f"{n}_ {i}")
```

#### reversed(iterable)

```python
myString = "Elzero"

print(reversed(myString))
for letter in reversed(myString):

  print(letter)

for s in reversed(mySkills):

  print(s)
```
