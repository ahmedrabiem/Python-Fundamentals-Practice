#### Lambda Func

###### [1] It Has No Name
###### [2] You Can Call It Inline Without Defining It
###### [3] You Can Use It In Return Data From Another Function
###### [4] Lambda Used For Simple Functions and Def Handle The Large Tasks
###### [5] Lambda is One Single Expression not Block Of Code
###### [6] Lambda Type is Function

```python
def say(name,age): return f"Hello {name} your age is {age}"
hello = lambda name,age: f"Hello {name} your age is {age}"

print(say("ahmed",20))
print(hello("ahmed",15))

print(say.__name__)
print(hello.__name__)
print(type(hello))
```

#### Map
##### [1] Map Take A Function + Iterator
##### [2] Map Called Map Because It Map The Function On Every Element
##### [3] The Function Can Be Pre-Defined Function or Lambda Function

##### 🚀 ملخص سريع عن دالة map() في بايثون

##### ✅ map(): دالة مدمجة تطبق دالة على كل عنصر في قائمة أو مجموعة

##### ✅ تُعيد: كائن map (يجب تحويله إلى list() أو tuple())

##### ✅ تُستخدم مع: دوال عادية أو lambda

##### ✅ مثال:
##### map(lambda x: x ** 2, [1, 2, 3]) → [1, 4, 9]


###### Use Map With Predefined Function

```python
def f (name):
    return f"- {name.strip().capitalize()} -"
l =["Ahmed","rabie","    saYed"]
myf=map(f,l)
print(myf)
for n in myf:
    print(n)

print(type(myf))
print("#"* 8)

for n in map(f,l):
    print(n)
```

###### Use Map With Lambda Function

```python
l =["Ahmed","rabie","    saYed"]

print(map(lambda x: f"- {x} -",l))

for n in map(lambda x: f"- {x.strip().capitalize()} -",l):
    print(n)
```

```python
numbers = [-10, 25, -30, 15, -5]

print(map(lambda n : abs(n),numbers))
for x in list(map(lambda n : abs(n),numbers)):
    print(x)
```

```python
numbers = [-10, 25, -30, 15, -5]

# تحويل كل عنصر إلى قيمته المطلقة باستخدام map()
absolute_numbers = list(map(abs, numbers))  

# طباعة النتيجة
print(absolute_numbers)
```

```python
students = ["Ali", "Sara", "Omar", "Lina"]
scores = [85, 92, 78, 90]
l=map(lambda name, score: {"name": name, "score": score}, students, scores)

for student in l:
    print(f"Name: {student['name']}, Score: {student['score']}")

```

##### -- Built In Functions => Filter --

##### [1] Filter Take A Function + Iterator
##### [2] Filter Run A Function On Every Element
##### [3] The Function Can Be Pre-Defined Function or Lambda Function
##### [4] Filter Out All Elements For Which The Function Return True
##### [5] The Function Need To Return Boolean Value

```python
def num(n):
    # if n>10
    # return n
    return n>10

mynum=[10,2,34,432,3,4,5,23]
print(filter(num,mynum))
for i in filter(num,mynum):
    print(i)

print("*"*5)

x = list(filter(num,mynum))
print(x)
for n in x:
    print(n)
```

```python
mynum=[10,2,34,432,3,4,5,23]
x = list(filter(lambda n: n>10,mynum))

for i in x:
    print(i)
```

```python
names = ["ahmed","sayed","awad","anas","yasser"]

v = list(filter(lambda name: name.startswith("a"),names))
print(v)

print("*"*25)

def finam(n):
    return n.startswith("a")

x=list(filter(finam,names))

print(x)
```

##### -- Built In Functions => Reduce --
####
##### [1] Reduce Take A Function + Iterator
##### [2] Reduce Run A Function On FIrst and Second Element And Give Result
##### [3] Then Run Function On Result And Third Element
##### [4] Then Run Function On Rsult And Fourth Element And So On
##### [5] Till One ELement is Left And This is The Result of The Reduce
##### [6] The Function Can Be Pre-Defined Function or Lambda Function

```python
from functools import reduce

num=[1,3,4,56,45]
result = reduce(lambda s1,s2: s1+s2  , num)
print(result)

print("*"*4)

def sumAll(num1, num2):
  return num1 + num2

numbers = [1 ,3,4,56,45]
result = reduce(sumAll, numbers)
print(result)
```
